# 01 - EDA Clientes

Análise exploratória da base de clientes do PI4.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# carregar base
df = pd.read_excel('../cadastros_08-052026 22_42_48.xls')

df.head()
```

```
# visão geral
print(df.shape)
print(df.info())
```

```
# estatísticas descritivas
df.describe(include='all').T
```

```
# valores nulos
nulls = (df.isnull().mean() * 100).sort_values(ascending=False)

plt.figure(figsize=(12,6))
nulls.head(20).plot(kind='bar')
plt.title('Top 20 campos com mais valores nulos')
plt.ylabel('% nulo')
plt.show()
```

```
# distribuição de perfis
perfil_col = 'Perfil'  # ajustar caso necessário

df[perfil_col].value_counts().head(10).plot(kind='bar')

plt.title('Distribuição de perfis')
plt.ylabel('Quantidade')
plt.show()
```

```
# análise temporal
data_col = 'Data Cadastro'  # ajustar nome

df[data_col] = pd.to_datetime(df[data_col], errors='coerce')

cadastros_mes = df.groupby(df[data_col].dt.to_period('M')).size()

cadastros_mes.plot(figsize=(12,5))

plt.title('Cadastros por mês')
plt.ylabel('Quantidade')
plt.show()
```

```
# duplicados
duplicados = df.duplicated().sum()

print(f'Duplicados encontrados: {duplicados}')
```
