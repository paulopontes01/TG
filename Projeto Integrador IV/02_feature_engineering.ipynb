# 02 - Feature Engineering

```
import pandas as pd
import numpy as np

df = pd.read_excel('../cadastros_08-052026 22_42_48.xls')

df.head()
```

```
# limpeza básica
df.columns = df.columns.str.strip()

# remover duplicados
df = df.drop_duplicates()

df.shape
```

```
# exemplo de criação de score de cliente

def gerar_score(row):
    score = 0

    if pd.notnull(row.get('Telefone')):
        score += 1

    if pd.notnull(row.get('Email')):
        score += 1

    if pd.notnull(row.get('Cidade')):
        score += 1

    return score

df['score_cliente'] = df.apply(gerar_score, axis=1)

df['categoria_cliente'] = pd.cut(
    df['score_cliente'],
    bins=[-1,1,2,3],
    labels=['baixo','medio','alto']
)

df[['score_cliente','categoria_cliente']].head()
```

```
# encoding
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df['target'] = encoder.fit_transform(df['categoria_cliente'])

df[['categoria_cliente','target']].head()
```

```
# salvar base tratada
df.to_csv('../data_clientes_tratada.csv', index=False)
```
